# 1. Load Dataset

Đọc bộ dữ liệu khảo sát sinh viên Việt Nam từ file CSV và kiểm tra dữ liệu đầu vào trước khi thực hiện các bước phân tích tổng quan.

Giai đoạn này chỉ thực hiện tải dữ liệu và quan sát ban đầu. Không thực hiện bất kỳ thao tác thay đổi hoặc xử lý dữ liệu.

```python
# Khai báo đường dẫn bộ dữ liệu khảo sát

dataset_path = Path("../../data/raw/vietnam_dataset/vn_digital_burnout.csv")


# Đọc bộ dữ liệu khảo sát

survey_dataset = pd.read_csv(dataset_path)


# Hiển thị thông tin cơ bản của bộ dữ liệu

print("Đã tải bộ dữ liệu khảo sát thành công.")

print(f"Số lượng quan sát: {survey_dataset.shape[0]:,}")

print(f"Số lượng biến: {survey_dataset.shape[1]}")


# Hiển thị một số dòng dữ liệu đầu tiên

display(survey_dataset.head())
```

# 2. Dataset Overview

Khảo sát tổng quan cấu trúc của bộ dữ liệu khảo sát sinh viên Việt Nam nhằm xác định quy mô, kiểu dữ liệu và đặc điểm của từng biến trước khi thực hiện các bước tiền xử lý.

- Kiểm tra thông tin tổng quan của bộ dữ liệu.
- Xác định số lượng quan sát và số lượng biến.
- Phân loại kiểu dữ liệu của từng biến.

```python
# Hiển thị thông tin tổng quan của bộ dữ liệu

print("Thông tin tổng quan của bộ dữ liệu khảo sát:")

survey_dataset.info()
```

```python
# Thống kê kích thước bộ dữ liệu

dataset_shape = pd.DataFrame(
    {
        "Metric": [
            "Number of Observations",
            "Number of Variables"
        ],
        "Value": [
            survey_dataset.shape[0],
            survey_dataset.shape[1]
        ]
    }
)

print("Kích thước bộ dữ liệu:")

display(dataset_shape)
```

```python
# Liệt kê các biến theo từng kiểu dữ liệu

data_type_table = pd.DataFrame(
    {
        "Variable": survey_dataset.columns,
        "Data Type": survey_dataset.dtypes.astype(str).values
    }
)

print("Danh sách kiểu dữ liệu của từng biến:")

display(data_type_table)
```

```python
# Phân nhóm các biến theo kiểu dữ liệu

object_variables = (
    survey_dataset
    .select_dtypes(include="object")
    .columns
    .tolist()
)

numeric_variables = (
    survey_dataset
    .select_dtypes(include=["number"])
    .columns
    .tolist()
)

print(f"Số biến định tính: {len(object_variables)}")

print(f"Số biến định lượng: {len(numeric_variables)}")
```

# 3. Data Quality Assessment

Đánh giá chất lượng ban đầu của bộ dữ liệu khảo sát sinh viên Việt Nam nhằm phát hiện các vấn đề có thể ảnh hưởng đến quá trình phân tích và xây dựng mô hình học máy.

- Kiểm tra giá trị thiếu.
- Kiểm tra bản ghi trùng lặp.
- Thống kê số lượng giá trị khác nhau.
- Thống kê mô tả dữ liệu.

## 3.1 Missing Values

Kiểm tra sự tồn tại của các giá trị thiếu trong bộ dữ liệu.

```python
# Thống kê số lượng và tỷ lệ giá trị thiếu

missing_summary = pd.DataFrame({

    "Variable": survey_dataset.columns,

    "Missing Count": survey_dataset.isna().sum().values,

    "Missing Percentage (%)": (
        survey_dataset.isna().mean() * 100
    ).round(2)

})

missing_summary = missing_summary.sort_values(

    by="Missing Count",
    ascending=False

)

print("Kết quả kiểm tra giá trị thiếu:")

display(missing_summary)
```

## 3.2 Duplicate Records

Kiểm tra sự tồn tại của các bản ghi bị trùng lặp trong bộ dữ liệu khảo sát.

```python
# Thống kê bản ghi trùng lặp

duplicate_count = survey_dataset.duplicated().sum()

duplicate_percentage = (

    duplicate_count
    / len(survey_dataset)

) * 100


duplicate_summary = pd.DataFrame({

    "Metric": [
        "Duplicate Records",
        "Duplicate Percentage (%)"
    ],

    "Value": [
        duplicate_count,
        round(duplicate_percentage, 2)
    ]

})

print("Kết quả kiểm tra bản ghi trùng lặp:")

display(duplicate_summary)
```

## 3.3 Unique Values

Kiểm tra số lượng giá trị duy nhất của từng biến nhằm hiểu rõ đặc điểm của các biến trong dataset.

```python
# Thống kê số lượng giá trị khác nhau

unique_summary = pd.DataFrame({

    "Variable": survey_dataset.columns,
    "Unique Values": survey_dataset.nunique().values

})

unique_summary = unique_summary.sort_values(

    by="Unique Values",
    ascending=False

)

print("Thống kê số lượng giá trị khác nhau:")

display(unique_summary)
```

## 3.4 Descriptive Statistics

Phân tích thống kê mô tả nhằm quan sát đặc điểm phân bố ban đầu của các biến trong dataset.

```python
# Thống kê mô tả các biến định lượng

numeric_summary = survey_dataset.describe().T

print("Thống kê mô tả các biến định lượng:")

display(numeric_summary)
```

```python
# Thống kê các biến định tính

categorical_summary = pd.DataFrame({

    "Variable": object_variables,

    "Unique Values": [
        survey_dataset[column].nunique()
        for column in object_variables
    ],

    "Most Frequent Value": [
        survey_dataset[column].mode().iloc[0]
        for column in object_variables
    ]

})

print("Thống kê các biến định tính:")

display(categorical_summary)
```

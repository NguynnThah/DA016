# 3. Data Quality Assessment

Đánh giá chất lượng dữ liệu ban đầu của bộ dữ liệu khảo sát sinh viên Việt Nam nhằm xác định các vấn đề tiềm ẩn trước khi thực hiện bước Cleaning & Preprocessing.

Phần này thực hiện đánh giá:

- Giá trị thiếu (Missing Values).
- Bản ghi trùng lặp (Duplicate Records).
- Số lượng giá trị duy nhất của từng biến (Unique Values).
- Thống kê mô tả dữ liệu (Descriptive Statistics).

Các kết quả thu được chỉ nhằm mục đích khảo sát và ghi nhận vấn đề.

## 3.1 Missing Values

Kiểm tra sự tồn tại của các giá trị thiếu trong từng biến nhằm đánh giá mức độ đầy đủ của dữ liệu khảo sát.

```python
# Kiểm tra số lượng và tỷ lệ giá trị thiếu trong từng biến

missing_summary = survey_dataset.isnull().sum().to_frame()

missing_summary.columns = ["missing_count"]

missing_summary["missing_percentage"] = (
    missing_summary["missing_count"]
    / len(survey_dataset)
) * 100


missing_summary = missing_summary.sort_values(
    by="missing_count",
    ascending=False
)

missing_summary
```

## 3.2 Duplicate Records

Kiểm tra sự tồn tại của các bản ghi bị trùng lặp trong bộ dữ liệu khảo sát.

```python
# Kiểm tra số lượng bản ghi trùng lặp

duplicate_records = survey_dataset.duplicated().sum()

print(f"Số lượng bản ghi trùng lặp: {duplicate_records}")
```

## 3.3 Unique Values

Kiểm tra số lượng giá trị duy nhất của từng biến nhằm hiểu rõ đặc điểm của các biến trong dataset.

```python
# Kiểm tra số lượng giá trị duy nhất của từng biến

unique_values = survey_dataset.nunique().to_frame()

unique_values.columns = ["unique_count"]


unique_values = unique_values.sort_values(
    by="unique_count",
    ascending=True
)

unique_values
```

## 3.4 Descriptive Statistics

Phân tích thống kê mô tả nhằm quan sát đặc điểm phân bố ban đầu của các biến trong dataset.

```python
# Thống kê mô tả cho các biến số

numerical_statistics = survey_dataset.describe()

numerical_statistics
```

```python
# Thống kê mô tả cho các biến phân loại

categorical_statistics = survey_dataset.describe(
    include="object"
)

categorical_statistics
```

## 4.1 Burnout Symptom Indicators

Kiểm tra các câu hỏi khảo sát liên quan đến biểu hiện Digital Burnout.
Các nhóm triệu chứng bao gồm:

- Emotional Exhaustion.
- Stress.
- Mental Fatigue.

```python
# Xác định các biến liên quan đến burnout symptoms

burnout_indicator_keywords = [
    "bã người",
    "kiệt sức",
    "căng thẳng",
    "lo lắng",
    "đầu nặng",
    "mệt mỏi"
]


burnout_indicator_columns = []

for column in survey_dataset.columns:
    for keyword in burnout_indicator_keywords:
        if keyword.lower() in column.lower():
            burnout_indicator_columns.append(column)


print("Các biến liên quan đến burnout symptoms:")

for column in burnout_indicator_columns:
    print(column)
```

## 4.4 Context Variables

Khảo sát các biến bối cảnh cá nhân và môi trường học tập nhằm mô tả đặc điểm của nhóm sinh viên tham gia khảo sát.
Các nhóm biến bao gồm:

- Thông tin cá nhân:
    - Giới tính.
    - Năm sinh.

- Thông tin học tập:
    - Giai đoạn học tập hiện tại.
    - Hình thức học tập/làm việc.

- Bối cảnh sử dụng thiết bị:
    - Mục đích sử dụng thiết bị số.

```python
# Xác định các biến thuộc nhóm context variables

context_keywords = [
    "Giới tính",
    "Năm sinh",
    "giai đoạn",
    "Hình thức",
    "Mô tả cách"
]


context_columns = []

for column in survey_dataset.columns:
    for keyword in context_keywords:
        if keyword.lower() in column.lower():
            context_columns.append(column)


print("Các biến thuộc nhóm Context Variables:")

for column in context_columns:
    print(column)
```
